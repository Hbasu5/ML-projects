```python
import pandas as pd
df = pd.read_csv("electricity_data.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.columns
```

```python
X = df[['Temperature']]
y = df[['Power_Usage']]
```

```python
plt.scatter(X,y)
plt.xlabel("TEMPERATURE")
plt.ylabel("POWER USAGE")
plt.title("TEMPERATURE V/S POWER USAGE")
plt.savefig("temp_vs_pwr_usage")
plt.show()
```

```python
X1 = df[['Appliances']]
y = df[['Power_Usage']]
```

```python
plt.scatter(X1,y)
plt.xlabel("APPLIANCES")
plt.ylabel("POWER USAGE")
plt.title("APPLIANCES V/S POWER USAGE")
plt.savefig("appliances_vs_pwr_usage")
plt.show()
```

```python
df.corr()
```

```python
sns.heatmap(df.corr(),annot=True,cmap="coolwarm")
plt.title("Correlation HeatMap")
plt.savefig("heatmap_correlation_seaborn")
plt.show()
```
