```python
import pandas as pd
df = pd.read_csv("cpu_data.csv")
df.head()
```

```python
df.columns
```

```python
X = df[['CPU_Usage_%']]
y = df[['CPU_Temp']]
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X,y)
plt.xlabel("CPU USAGE %")
plt.ylabel("CPU TEMP")
plt.title("CPU USAGE V/S CPU TEMP")
plt.savefig("cpu_usage_vs_cpu_temp")
plt.show()
```

```python
X = df[['Ambient_Temp']]
y = df[['CPU_Temp']]
plt.scatter(X,y)
plt.xlabel("AMBIENT TEMPATURE")
plt.ylabel("CPU TEMP")
plt.title("AMBIENT TEMPERATURE V/S CPU TEMP")
plt.savefig("ambient_temperature_vs_cpu_temp")
plt.show()
```
