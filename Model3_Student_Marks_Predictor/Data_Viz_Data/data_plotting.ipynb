```python
import pandas as pd
df = pd.read_csv("predictor_data.csv")
df.head()
```

```python
df.columns
```

```python
import matplotlib.pyplot as plt
features = ['Midterm_Score','Attendance_Pct','Study_Hours','Sleep_Hours','Distraction_Level','Projects']
y = df['Final_Marks']
for feature in features:
    plt.scatter(df[feature],y)
    plt.xlabel(feature.replace("_"," ").upper())
    plt.ylabel("FINAL MARKS")
    plt.title(f"{feature.replace('_',' ').title()} vs Final Marks")
    plt.savefig(feature.lower()+"_vs_final_marks")
    plt.show()
```
